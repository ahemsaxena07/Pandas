```python
a = pd.Series([np.nan, 3.5, 0.0, 3.5, 4.5, np.nan], index = ["f", "e", "d", "c", "b", "a"])
```

```python
b = pd.Series([0., np.nan, 2., np.nan, np.nan, 5.], index = ["a", "b", "c", "d", "e", "f"])
```

```python
a
```

```python
b
```

```python
np.where(pd.isna(a), b, a)
```

```python
a.combine_first(b)
```
