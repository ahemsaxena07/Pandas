```python
text = """Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@gmail.com"""
```

```python
pattern = r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}"
```

```python
regex = re.compile(pattern, flags = re.IGNORECASE)
```

```python
regex.findall(text)
```

```python
m = regex.search(text)
```

```python
m
```

```python
text[m.start():m.end()]
```

```python
print(regex.sub("REDATED", text))
```

```python
regex.findall(text)
```
