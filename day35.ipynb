```python
import numpy as np
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({'key': ['a', 'b', 'c'] * 4, 'value': np.arange(12.)})
```

```python
df
```

```python
g = df.groupby('key')['value']
```

```python
g.mean()
```

```python
def get_mean(group):
    return group.mean()
```

```python
g.transform(get_mean)
```

```python
g.transform('mean')
```

```python
def times_two(group):
    return group * 2
```

```python
g.transform(times_two)
```

```python
def get_ranks(group):
    return group.rank(ascending = False)
```

```python
g.transform(get_ranks)
```

```python
def normalize(x):
    return (x - x.mean()) / x.std()
```

```python
g.transform(normalize)
```

```python
g.apply(normalize)
```

```python
g.transform('mean')
```

```python
normalized = (df['value'] - g.transform('mean')) / g.transform('std')
```
