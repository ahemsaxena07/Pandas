```python
s1 = pd.Series([0,1], index = ["a", "b"], dtype = "Int64")
```

```python
s2 = pd.Series([2,3,4], index = ["c", "d", "e"], dtype = "Int64")
```

```python
s3 = pd.Series([5, 6], index = ["f", "g"], dtype = "Int64")
```

```python
s1 
```

```python
s2
```

```python
s3
```

```python
pd.concat([s1, s2, s3])
```

```python
pd.concat([s1,s2,s3], axis = 1)
```

```python
s4 = pd.concat([s1,s3])
```

```python
s4
```

```python
pd.concat([s1,s4], axis = 1)
```

```python
pd.concat([s1,s4], axis=1, join="inner")
```
