```python
ages = [20, 22, 25, 27, 21, 23, 37, 31,  61, 45, 41, 32]
```

```python
bins = [18, 25, 35, 60, 100]
```

```python
age_categories = pd.cut(ages, bins)
```

```python
age_categories
```

```python
age_categories.codes
```

```python
age_categories.categories
```

```python
age_categories.categories[0]
```

```python
age_categories.value_counts()
```

```python
marks = [ 30, 54, 69, 100]
```

```python
bins = [0,45, 70, 100]
```

```python
marks_categories = pd.cut(marks, bins, labels = ["fail", "pass", "good"])
```

```python
marks_categories
```

```python
marks_categories.value_counts()
```

```python
pd.cut(marks, bins, right=False)
```
