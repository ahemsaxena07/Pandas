```python
from datetime import datetime
```

```python
from pandas.tseries.offsets import Day, MonthEnd
```

```python
now = datetime(2011, 11, 17)
```

```python
now + 3 * Day()
```

```python
now + MonthEnd()
```

```python
now + MonthEnd(2)
```

```python
offset = MonthEnd()
```

```python
offset.rollforward(now)
```

```python
offset.rollback(now)
```
